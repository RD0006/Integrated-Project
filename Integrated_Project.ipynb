```python
from pathlib import Path
import kagglehub

# 1. Select the scene
SCENE_NAME = "treehill"

# 2. Locate or download the dataset
DATASET = Path(kagglehub.dataset_download("thnhdg/testing"))

# 3. Locate the scene folder
SCENE = DATASET / "360_v2" / SCENE_NAME

print("Dataset:", DATASET)
print("Dataset exists:", DATASET.exists())
print("Scene:", SCENE)
print("Scene exists:", SCENE.exists())

# 4. If the expected path doesn't exist, search for the scene
if not SCENE.exists():
    matches = [
        p for p in DATASET.rglob(SCENE_NAME)
        if p.is_dir() and p.name == SCENE_NAME
    ]

    if matches:
        SCENE = matches[0]
        print("Found scene at:", SCENE)
    else:
        raise FileNotFoundError(
            f"Scene '{SCENE_NAME}' was not found inside {DATASET}"
        )

# 5. Check the images folder
IMAGES = SCENE / "images"

if IMAGES.is_dir():
    image_files = [
        p for p in IMAGES.iterdir()
        if p.is_file() and p.suffix.lower() in
        {".jpg", ".jpeg", ".png", ".webp"}
    ]
    print("Images found:", len(image_files))
else:
    image_files = []
    print("Images folder not found:", IMAGES)

# 6. Check COLMAP sparse reconstruction files
SPARSE = SCENE / "sparse" / "0"
required_files = ["cameras.bin", "images.bin", "points3D.bin"]

print("\nCOLMAP file status:")
for name in required_files:
    file = SPARSE / name
    print(f"{name}: {'OK' if file.is_file() else 'MISSING'}")

# 7. Validate
assert image_files, (
    f"No images found in {IMAGES}. "
    "Inspect the scene folder structure."
)

missing = [name for name in required_files if not (SPARSE / name).is_file()]

if missing:
    print("\nCOLMAP reconstruction files are missing:", missing)
    print("Scene contents:")
    for p in SCENE.iterdir():
        print(" -", p.name)
    print(
        "\nYou need to locate or generate the COLMAP sparse reconstruction "
        "before training with this scene."
    )
else:
    print("\nSUCCESS: Images and COLMAP files are ready for training.")
```

```python
# ============================================================
# 3D Gaussian Splatting: Evaluate all trained scenes
# Excludes: room, treehill
# Metrics: PSNR, SSIM, LPIPS, iterations, resolution
# ============================================================

from pathlib import Path
import subprocess
import json
import re
import ast
import pandas as pd
from IPython.display import display

# ---------------- CONFIGURATION ----------------
DRIVE_ROOT = Path("/content/drive/MyDrive/GaussianSplatting")
REPO = Path("/content/gaussian-splatting")
METRICS_SCRIPT = REPO / "metrics.py"

EXCLUDED_SCENES = {"room", "treehill"}

# Set the actual training downsampling factor for each scene if known.
# Example: {"bonsai": 4, "bicycle": 8, "counter": 4}
# Leave empty to infer it from saved configuration where possible.
RESOLUTION_OVERRIDES = {}

# ---------------- VALIDATE PATHS ----------------
if not DRIVE_ROOT.exists():
    raise FileNotFoundError(
        f"Google Drive folder not found: {DRIVE_ROOT}\n"
        "Mount Drive and update DRIVE_ROOT if necessary."
    )

if not METRICS_SCRIPT.exists():
    raise FileNotFoundError(
        f"metrics.py not found: {METRICS_SCRIPT}\n"
        "Update REPO to your 3DGS repository location."
    )

# ---------------- HELPER FUNCTIONS ----------------
def find_model_folder(scene_dir):
    """Find the highest-iteration saved model folder."""
    candidates = [
        p for p in scene_dir.rglob("iterations_*")
        if p.is_dir()
        and re.fullmatch(r"iterations_\d+", p.name)
    ]

    if candidates:
        return max(candidates, key=lambda p: int(p.name.split("_")[-1]))

    # Also support standard 3DGS output folders containing point_cloud.
    standard_candidates = list(scene_dir.glob("point_cloud/iteration_*"))
    standard_candidates = [
        p for p in standard_candidates
        if p.is_dir() and (p / "point_cloud.ply").exists()
    ]

    if standard_candidates:
        return max(
            standard_candidates,
            key=lambda p: int(p.name.split("_")[-1])
        )

    return None


def get_iterations(model_dir):
    """Extract the iteration count from the folder name."""
    match = re.search(r"(?:iterations_|iteration_)(\d+)$", model_dir.name)
    return int(match.group(1)) if match else None


def get_resolution(scene_name, model_dir):
    """
    Return the training downsampling factor when known.
    cfg_args resolution=-1 means automatic sizing and does not
    tell us whether the training images were downsampled by 4 or 8.
    """
    if scene_name in RESOLUTION_OVERRIDES:
        return RESOLUTION_OVERRIDES[scene_name]

    possible_cfgs = [
        model_dir / "cfg_args",
        model_dir.parent / "cfg_args",
        model_dir.parent.parent / "cfg_args",
        DRIVE_ROOT / scene_name / "cfg_args",
    ]

    for cfg in possible_cfgs:
        if not cfg.is_file():
            continue

        text = cfg.read_text(errors="ignore")

        # Standard 3DGS cfg_args often contains Namespace(...).
        match = re.search(r"\bresolution\s*=\s*(-?\d+)", text)
        if match:
            value = int(match.group(1))
            return "Auto (-1)" if value == -1 else value

    return "Unknown"


def find_evaluation_path(model_dir, scene_dir):
    """Find the folder to pass to metrics.py."""
    # The official metrics.py normally expects a model output directory
    # containing point_cloud/iteration_XXXX/point_cloud.ply.
    if (model_dir / "point_cloud").is_dir():
        return model_dir

    if model_dir.name.startswith("iteration_"):
        # Model path is .../point_cloud/iteration_XXXX/
        output_dir = model_dir.parent.parent
        if (output_dir / "point_cloud").is_dir():
            return output_dir

    # Custom iterations_5000 / iterations_7000 folders may themselves
    # be model output folders or contain a standard point_cloud folder.
    if (model_dir / "point_cloud" / "point_cloud.ply").exists():
        return model_dir

    return None


# ---------------- EVALUATE MODELS ----------------
results = []

scene_dirs = sorted(
    p for p in DRIVE_ROOT.iterdir()
    if p.is_dir() and p.name.lower() not in EXCLUDED_SCENES
)

for scene_dir in scene_dirs:
    scene_name = scene_dir.name
    model_dir = find_model_folder(scene_dir)

    if model_dir is None:
        print(f"SKIP {scene_name}: no saved iteration folder found.")
        continue

    iterations = get_iterations(model_dir)
    resolution = get_resolution(scene_name, model_dir)
    eval_path = find_evaluation_path(model_dir, scene_dir)

    print(f"\n{'=' * 60}")
    print(f"Scene      : {scene_name}")
    print(f"Model      : {model_dir}")
    print(f"Iterations : {iterations}")
    print(f"Resolution : {resolution}")

    if eval_path is None:
        print("SKIP: Model folder does not match the expected metrics.py structure.")
        results.append({
            "Scene": scene_name,
            "Iterations": iterations,
            "Resolution": resolution,
            "PSNR": None,
            "SSIM": None,
            "LPIPS": None,
            "Status": "Unsupported model folder structure",
        })
        continue

    try:
        proc = subprocess.run(
            ["python", str(METRICS_SCRIPT), str(eval_path)],
            cwd=str(REPO),
            capture_output=True,
            text=True,
            timeout=3600,
        )

        print(proc.stdout[-3000:])

        if proc.returncode != 0:
            print("Evaluation error:\n", proc.stderr[-2000:])
            results.append({
                "Scene": scene_name,
                "Iterations": iterations,
                "Resolution": resolution,
                "PSNR": None,
                "SSIM": None,
                "LPIPS": None,
                "Status": "Evaluation failed",
            })
            continue

        # Search common locations for metrics JSON output.
        json_candidates = [
            eval_path / "results.json",
            eval_path / "results" / "results.json",
            REPO / "results.json",
        ]

        # Some versions write metrics.json instead.
        json_candidates += [
            eval_path / "metrics.json",
            eval_path / "results" / "metrics.json",
        ]

        metric_data = None

        for json_path in json_candidates:
            if json_path.exists():
                try:
                    loaded = json.loads(json_path.read_text())
                    if isinstance(loaded, dict):
                        metric_data = loaded
                        break
                except (json.JSONDecodeError, OSError):
                    pass

        # Look for metric values in stdout if no JSON was found.
        stdout = proc.stdout

        def extract_metric(data, name):
            if data:
                # Handle both direct metrics and nested scene results.
                if name in data and isinstance(data[name], (int, float)):
                    return data[name]

                for value in data.values():
                    if isinstance(value, dict) and name in value:
                        try:
                            return float(value[name])
                        except (TypeError, ValueError):
                            pass

            pattern = rf"\b{re.escape(name)}\b\s*[:=]\s*(-?\d+(?:\.\d+)?)"
            match = re.search(pattern, stdout, re.IGNORECASE)
            return float(match.group(1)) if match else None

        psnr = extract_metric(metric_data, "PSNR")
        ssim = extract_metric(metric_data, "SSIM")
        lpips = extract_metric(metric_data, "LPIPS")

        status = (
            "Completed"
            if all(v is not None for v in [psnr, ssim, lpips])
            else "Ran; metrics need verification"
        )

        results.append({
            "Scene": scene_name,
            "Iterations": iterations,
            "Resolution": resolution,
            "PSNR": psnr,
            "SSIM": ssim,
            "LPIPS": lpips,
            "Status": status,
        })

    except subprocess.TimeoutExpired:
        print("Evaluation timed out.")
        results.append({
            "Scene": scene_name,
            "Iterations": iterations,
            "Resolution": resolution,
            "PSNR": None,
            "SSIM": None,
            "LPIPS": None,
            "Status": "Timed out",
        })


# ---------------- DISPLAY AND SAVE ----------------
df = pd.DataFrame(results)

if not df.empty:
    display(df)

    output_csv = DRIVE_ROOT / "evaluation_results.csv"
    df.to_csv(output_csv, index=False)
    print(f"\nResults saved to: {output_csv}")
else:
    print("No eligible trained models were found.")
```
